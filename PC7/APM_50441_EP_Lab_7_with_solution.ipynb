The *sampling trick* (lecture): for many questions, a random sample is enough, as the precision of an estimate depends on the size of the sample. Often, we do not need the full population to obtain a good enough estimate (PC7, *Reservoir sampling*).

**Question** Estimate the mean arrival delay of all the flights of the sample with random subsamples of $n = 100$, $1000$, $10\,000$ and $100\,000$ flights ($200$ subsamples each): compare the standard deviation of the estimates with $\sigma/\sqrt{n}$. Then compare a simple random subsample of $2004$ flights with a subsample *stratified* by month ($167$ flights each month, the monthly means being weighted by the proportions of the months).

**Answer**

````r
delays <- flights$ARRIVAL_DELAY[!is.na(flights$ARRIVAL_DELAY)]
set.seed(1)
precision <- map_dfr(c(100, 1000, 10000, 100000), function(n) {
  estimates <- replicate(200, mean(sample(delays, n)))
  tibble(n = n, `sd of the estimates` = sd(estimates), `sigma / sqrt(n)` = sd(delays) / sqrt(n),
         `finite population SE` = sd(delays) * sqrt((1 - n / length(delays)) / n))
})
precision
with_month <- flights |> filter(!is.na(ARRIVAL_DELAY)) |> select(MONTH, ARRIVAL_DELAY)
monthly_weights <- with_month |> count(MONTH) |> mutate(weight = n / sum(n))
stratified <- replicate(200, with_month |> group_by(MONTH) |> slice_sample(n = 167) |>
                          summarise(m = mean(ARRIVAL_DELAY)) |> left_join(monthly_weights, by = "MONTH") |>
                          summarise(sum(m * weight)) |> pull())
c(`simple random sampling` = sd(replicate(200, mean(sample(with_month$ARRIVAL_DELAY, 2004)))), stratified = sd(stratified))
````

**Conclusion** At $n=100{,}000$ the standard error for the fixed file mean is $0.1125$ minutes after correcting for sampling without replacement from $571{,}396$ observed delays. Inference about future flights requires further sampling assumptions. The independent-sampling benchmark is $\sigma/\sqrt{n}$ ($3.9$ for $n = 100$, $0.12$ for $n = 100\,000$): it depends on the size $n$ of the sample: we do not need the $5.8$ million flights to estimate the mean delay well. A sample of a few thousand flights gives the mean delay within a minute, enough to explore the data. Here, the stratification by month does not help ($0.85$ against $0.84$): the monthly means differ little compared with the huge variability of the delays within a month. Stratification pays off when the strata explain a large part of the variance.

**Answer**

````r
if (!dir.exists("data/flights_full_parquet")) {  # the full data, downloaded once (12 files of about 12 MB)
  for (m in 1:12) {
    file <- sprintf("data/flights_full_parquet/MONTH=%d/part-0.parquet", m)
    dir.create(dirname(file), recursive = TRUE, showWarnings = FALSE)
    download.file(paste0("https://raw.githubusercontent.com/APM-50441-EP/APM_50441_EP-2026-2027/main/PC7/", file), file, mode = "wb")
  }
}
options(arrow.use_altrep = FALSE)  # real R vectors (otherwise Arrow converts the columns lazily, when they are used)
start <- Sys.time()
flights_full <- open_dataset("data/flights_full_parquet") |> collect()
c(`loading time (s)` = as.numeric(difftime(Sys.time(), start, units = "secs")), `size (MB)` = as.numeric(lobstr::obj_size(flights_full)) / 1e6)

con <- dbConnect(duckdb())
by_airline_dplyr <- function(data) data |> group_by(AIRLINE) |>
  summarise(arrival_delay = mean(ARRIVAL_DELAY, na.rm = TRUE), late = mean(ARRIVAL_DELAY > 15, na.rm = TRUE))
by_airline_dt <- function(data_dt) data_dt[, .(arrival_delay = mean(ARRIVAL_DELAY, na.rm = TRUE),
                                               late = mean(ARRIVAL_DELAY > 15, na.rm = TRUE)), by = AIRLINE]
by_airline_duckdb <- function(data_db) data_db |> group_by(AIRLINE) |>
  summarise(arrival_delay = mean(ARRIVAL_DELAY, na.rm = TRUE), late = mean(as.numeric(ARRIVAL_DELAY > 15), na.rm = TRUE)) |>
  collect()
compare_engines <- function(data, name, iterations) {
  data_dt <- as.data.table(data)    # a copy
  duckdb_register(con, name, data)  # no copy: DuckDB reads the columns of the R data frame
  data_db <- tbl(con, name)
  results <- list(dplyr = by_airline_dplyr(data), data.table = by_airline_dt(data_dt), duckdb = by_airline_duckdb(data_db)) |>
    map(\(r) arrange(as_tibble(r), AIRLINE))
  agree <- c(`data.table agrees` = isTRUE(all.equal(results$dplyr, results$data.table, check.attributes = FALSE)),
             `duckdb agrees` = isTRUE(all.equal(results$dplyr, results$duckdb, check.attributes = FALSE)))
  timing <- bench::mark(dplyr = by_airline_dplyr(data), data.table = by_airline_dt(data_dt), duckdb = by_airline_duckdb(data_db),
                        iterations = iterations, check = FALSE, filter_gc = FALSE)
  list(results = results, agree = agree, timing = timing |> transmute(engine = as.character(expression), median, mem_alloc))
}
on_full <- compare_engines(flights_full, "flights_full", iterations = 5)
on_sample <- compare_engines(flights, "flights", iterations = 10)
rbind(`full data` = on_full$agree, `10% sample` = on_sample$agree)
bind_rows(`full data` = on_full$timing, `10% sample` = on_sample$timing, .id = "data") |>
  pivot_wider(names_from = data, values_from = c(median, mem_alloc)) |> print(width = Inf)
results <- on_full$results       # the exact statistics
flights_db <- tbl(con, "flights")  # the sample in DuckDB, used in the To go further part
left_join(results$dplyr |> select(AIRLINE, exact = arrival_delay),
          on_sample$results$dplyr |> select(AIRLINE, `10% sample` = arrival_delay), by = "AIRLINE")
````

**Conclusion** The three engines give the same results, on the sample as on the full data. Loading the $5.8$ million flights from the Parquet files takes a few seconds and $1.4$ GB of memory. ## 3. Profiling and optimizing

The cumulated arrival delay along the year can be computed with `cumsum`. To understand where the time goes, we write this function ourselves (PC7, *Complexity of cumulative sums*).

````r
cumsum_0 <- function(x) {
  res <- numeric(0)
  for (i in seq_along(x)) {
    res[i] <- 0
    for (j in seq_len(i)) res[i] <- res[i] + x[j]
  }
  res
}
````

**Question** *Predict before you run:* how does the time of `cumsum_0` grow with the length $n$ of `x`? Measure it for $n = 250, 500, 1000, 2000$ and plot it in log-log scale.

**Answer**

````r
scaling <- tibble(n = c(250, 500, 1000, 2000)) |>
  mutate(time = map_dbl(n, \(n) { x <- rnorm(n); as.numeric(bench::mark(cumsum_0(x), iterations = 3)$median) }))
scaling
ggplot(scaling, aes(n, time)) + geom_point() + geom_line() + scale_x_log10() + scale_y_log10() + labs(y = "time (s)")
coef(lm(log(time) ~ log(n), data = scaling))[2]
````

**Conclusion** The time of `cumsum_0` grows like $n^2$: the slope in log-log scale is close to $2$ (between $1.8$ and $2$ depending on the run, the small sizes being dominated by constant costs), as expected from the $n(n-1)/2$ additions of the double loop (PC7, *Complexity of cumulative sums*). For the $582\,000$ flights, it would take hours.

**Question** Write `cumsum_1` with a single loop (`res[i] <- res[i - 1] + x[i]`) growing the result vector, `cumsum_2` with a preallocated result (`numeric(length(x))`), and compare them with the built-in `cumsum` for $n = 10^5$. **Conclusion** With a single loop, the algorithm is linear, but its R implementation is still $10$ to $15$ times slower than the built-in `cumsum` when the result is preallocated, and about $40$ times slower when the vector grows at each step (which also allocates $8$ times more memory; recent versions of R grow vectors in place by blocks, which limits the damage compared with the theoretical $O(n^2)$ copies). The compiled C++ loop (`Rcpp`) is as fast as the built-in `cumsum` (itself written in C), and `std::partial_sum` is even a little faster. The order of the optimizations matters: first the algorithm ($O(n^2) \to O(n)$), then the memory management, then the language.

## 4. Parallel cross validation

We want to predict the arrival delay from the information available at the departure. We compare $5$ linear models by an $8$-fold cross validation (note: in a formula, `DEPARTURE_DELAY/SCHEDULED_TIME` means *nesting*, not a ratio: the ratio is written `I(DEPARTURE_DELAY / SCHEDULED_TIME)`).

````r
flights_model_data <- flights |> select(ARRIVAL_DELAY, DEPARTURE_DELAY, SCHEDULED_TIME, AIRLINE) |> drop_na()
formulas <- c("ARRIVAL_DELAY ~ DEPARTURE_DELAY",
              "ARRIVAL_DELAY ~ DEPARTURE_DELAY + SCHEDULED_TIME",
              "ARRIVAL_DELAY ~ DEPARTURE_DELAY + SCHEDULED_TIME + I(DEPARTURE_DELAY / SCHEDULED_TIME)",
              "ARRIVAL_DELAY ~ DEPARTURE_DELAY + SCHEDULED_TIME + I(DEPARTURE_DELAY / SCHEDULED_TIME) + AIRLINE",
              "ARRIVAL_DELAY ~ (DEPARTURE_DELAY + SCHEDULED_TIME + I(DEPARTURE_DELAY / SCHEDULED_TIME)) * AIRLINE")
set.seed(2)
folds <- sample(rep(1:8, length.out = nrow(flights_model_data)))
cv_rmse <- function(formula, fold, data, folds) {
  fit <- lm(as.formula(formula), data = data[folds != fold, ])
  sqrt(mean((predict(fit, data[folds == fold, ]) - data$ARRIVAL_DELAY[folds == fold])^2))
}
tasks <- expand_grid(formula = formulas, fold = 1:8)
````

**Question** Compute the $40$ cross validation errors sequentially with `map2_dbl`, and display the mean CV RMSE of each model. Which model would you choose?

**Answer**

````r
sequential_time <- system.time(
  tasks$rmse <- map2_dbl(tasks$formula, tasks$fold, cv_rmse, data = flights_model_data, folds = folds)
)[["elapsed"]]
tasks |> group_by(formula) |> summarise(`CV RMSE` = mean(rmse), `fold SD` = sd(rmse)) |>
  mutate(model = match(formula, formulas), .before = 1) |> arrange(model) |> select(-formula)
paired <- tasks |> mutate(model = match(formula, formulas)) |> filter(model %in% 4:5) |>
  select(fold, model, rmse) |> pivot_wider(names_from = model, values_from = rmse, names_prefix = "m")
summary(paired$m5 - paired$m4)
sequential_time
````

### Map/Reduce by hand (★★)

In the Map/Reduce model, each partition of the data computes a small summary (*map*) and the summaries are merged (*reduce*). The least squares estimate only needs $\mathbb{X}^\top\mathbb{X}$ and $\mathbb{X}^\top\mathbf{Y}$, which are sums over the partitions (PC7, *Map/Reduce for statistics*).

**Question** For each monthly partition of the Parquet dataset, compute the number of flights, the mean and the sum of squared deviations of the arrival delay, and the matrices $\mathbb{X}_k^\top\mathbb{X}_k$ and $\mathbb{X}_k^\top\mathbf{Y}_k$ for the model `ARRIVAL_DELAY ~ DEPARTURE_DELAY + SCHEDULED_TIME`. Merge them, and compare the mean, the variance and the coefficients with the computations on the whole data.

**Answer**

````r
summaries <- map(1:12, function(m) {
  part <- open_dataset("data/flights_parquet") |> filter(MONTH == !!m) |>
    select(ARRIVAL_DELAY, DEPARTURE_DELAY, SCHEDULED_TIME) |> collect() |> drop_na()
  X <- cbind(1, part$DEPARTURE_DELAY, part$SCHEDULED_TIME)
  list(n = as.numeric(nrow(part)),                         # a double: n_A * n_B would overflow the integers
       mean = mean(part$ARRIVAL_DELAY), M = sum((part$ARRIVAL_DELAY - mean(part$ARRIVAL_DELAY))^2),
       XtX = crossprod(X), XtY = crossprod(X, part$ARRIVAL_DELAY))
})
merged <- reduce(summaries, function(a, b) {
  n <- a$n + b$n
  list(n = n, mean = (a$n * a$mean + b$n * b$mean) / n, M = a$M + b$M + a$n * b$n / n * (a$mean - b$mean)^2,
       XtX = a$XtX + b$XtX, XtY = a$XtY + b$XtY)
})
whole <- flights |> select(ARRIVAL_DELAY, DEPARTURE_DELAY, SCHEDULED_TIME) |> drop_na()
rbind(`map/reduce` = c(mean = merged$mean, variance = merged$M / (merged$n - 1), solve(merged$XtX, merged$XtY)[, 1]),
      `whole data` = c(mean(whole$ARRIVAL_DELAY), var(whole$ARRIVAL_DELAY), coef(lm(ARRIVAL_DELAY ~ DEPARTURE_DELAY + SCHEDULED_TIME, data = whole))))
````

**Conclusion** The merged summaries give exactly the mean, the variance and the coefficients of the least squares regression computed on the whole data: the means and the sums of squares are merged with Chan's formula, and $\mathbb{X}^\top\mathbb{X}$ and $\mathbb{X}^\top\mathbf{Y}$ are sums over the partitions (PC7, *Map/Reduce for statistics*). Each month only sends a few numbers, whatever its size: this is how statistics are computed on distributed data. The count must be stored as a double: the products of counts overflow R's 32-bit integers.

### Larger than memory (★★)

DuckDB can compute on data larger than the memory it is allowed to use: it streams through the files in chunks, and writes its intermediate results to disk if needed. A median or a quantile is harder than a mean: it is not a mergeable statistic (PC7, *Map/Reduce*), so the values of each group have to be kept or sorted.

**Question** If the full CSV file `data/flights.csv.gz` is available (it is not downloaded automatically: $195$ MB), limit the memory of DuckDB to $500$ MB (`SET memory_limit = '500MB'`) and compute, directly from the compressed CSV file, the number of flights, the median and the $90\%$ quantile of the arrival delay of the $10$ busiest origin airports, with their names (a join with `data/airports.csv`). Measure the time and the memory of the R session.

**Answer**

````r
if (file.exists("data/flights.csv.gz")) {
  dbExecute(con, "SET memory_limit = '500MB'")
  dbExecute(con, "SET threads = 4")                         # fewer reading buffers at the same time
  dbExecute(con, "SET preserve_insertion_order = false")    # the order of the rows does not matter here
  dbExecute(con, sprintf("SET temp_directory = '%s'", gsub("\\\\", "/", file.path(tempdir(), "duckdb"))))  # where DuckDB spills
  memory_before <- lobstr::mem_used()
  start <- Sys.time()
  airports_delays <- dbGetQuery(con, "
    SELECT f.ORIGIN_AIRPORT AS airport, a.AIRPORT AS name, COUNT(*) AS flights,
           MEDIAN(f.ARRIVAL_DELAY) AS median_delay, QUANTILE_CONT(f.ARRIVAL_DELAY, 0.9) AS q90_delay
    FROM read_csv_auto('data/flights.csv.gz', types = {'ORIGIN_AIRPORT': 'VARCHAR', 'DESTINATION_AIRPORT': 'VARCHAR'}) f
    JOIN read_csv_auto('data/airports.csv') a ON f.ORIGIN_AIRPORT = a.IATA_CODE
    GROUP BY ALL ORDER BY flights DESC LIMIT 10")
  print(c(`time (s)` = as.numeric(difftime(Sys.time(), start, units = "secs")),
          `R memory change (MB)` = as.numeric(lobstr::mem_used() - memory_before) / 1e6))
  for (setting in c("memory_limit", "threads", "preserve_insertion_order")) dbExecute(con, paste("RESET", setting))
  as_tibble(airports_delays)
} else {
  message("data/flights.csv.gz is not available")
}
````

**Conclusion** DuckDB computes the median and the $90\%$ quantile of the delays of the busiest airports directly from the compressed CSV file in a few seconds, with at most $500$ MB of memory, and the memory of the R session does not change: it streams through the file, keeping only the two needed columns. Most flights arrive early or on time (the medians are between $-6$ and $-3$ minutes), but the tails differ: $10\%$ of the flights leaving Chicago O'Hare arrive more than $51$ minutes late, against $29$ minutes for Atlanta and Minneapolis. A median or a quantile is not a mergeable statistic: DuckDB keeps the values of each airport (and would spill them to disk if needed), while the means of Part 2 only need sums and counts.

### The memory hierarchy (★★)

R matrices are stored column by column: reading a column reads contiguous memory, reading a row jumps from one column to the next. The caches of the processor load memory by blocks (*cache lines*), so that contiguous accesses are much faster.

**Question** For a $5000 \times 5000$ matrix, compare the time to sum its columns one by one (`m[, j]`) and its rows one by one (`m[i, ]`). Compare also `colSums` and `rowSums`.

**Answer**

````r
m <- matrix(rnorm(5000 * 5000), 5000, 5000)
bench::mark(`columns one by one` = for (j in 1:5000) sum(m[, j]), `rows one by one` = for (i in 1:5000) sum(m[i, ]),
            colSums = colSums(m), rowSums = rowSums(m), iterations = 3, check = FALSE, filter_gc = FALSE) |>
  transmute(loop = as.character(expression), median)
rm(m)
````

**Conclusion** Summing the $5000$ rows one by one takes $2.5$ to $3$ times longer than summing the $5000$ columns one by one, for the same number of additions: R stores the matrices column by column, so that a row is scattered in memory and each of its elements needs a new cache line. The vectorized `colSums` and `rowSums` are much faster than the loops (about $40$ to $100$ ms), `colSums` being faster than `rowSums` for the same reason. The way data are laid out in memory matters as much as the number of operations.

### How many workers? (★)

**Question** Run the $40$ tasks of the cross validation of Part 4 with $1$, $2$, $4$ and $8$ `mirai` workers (`mirai::daemons(k)`; at most $4$ on a machine with few cores, such as Colab: each worker holds a copy of the data and of its fits, about $1$ GB here), and plot the speed-up against the number of workers, with the number of cores of your machine (`parallel::detectCores()`). Fit Amdahl's law $S(P) = 1/((1-f) + f/P)$ to the measured speed-ups. What limits the speed-up? (PC7, *Speed-up laws*.)

**Answer**

````r
cv_all <- function() map2_dbl(tasks$formula, tasks$fold, in_parallel(\(formula, fold) cv_rmse(formula, fold, data, folds),
                                                                       cv_rmse = cv_rmse, data = flights_model_data, folds = folds))
n_workers <- c(1, 2, 4, 8)
n_workers <- n_workers[n_workers <= max(4, parallel::detectCores())]  # 8 workers would exhaust the memory of Colab
speed_up <- map_dfr(n_workers, function(k) {
  mirai::daemons(k)
  cv_all()                                                 # a first call sends the data to the workers
  elapsed <- system.time(cv_all())[["elapsed"]]
  mirai::daemons(0)
  tibble(workers = k, time = elapsed)
}) |> mutate(`speed-up` = time[1] / time)
amdahl <- nls(`speed-up` ~ 1 / ((1 - f) + f / workers), data = speed_up, start = list(f = 0.9))
speed_up
c(cores = parallel::detectCores(), `parallel fraction f` = coef(amdahl)[["f"]])
ggplot(speed_up, aes(workers, `speed-up`)) + geom_point() + geom_line() +
  geom_line(aes(y = predict(amdahl)), linetype = "dashed") + geom_abline(linetype = "dotted") + scale_x_log10() + scale_y_log10()
````

**Conclusion** The speed-up grows with the number of workers but less and less: about $1.8$ with $2$ workers, $2.4$ to $3$ with $4$ and $2.8$ to $3.6$ with $8$ (depending on the run and on the load of the machine), far from linear, although the machine has $22$ cores. Amdahl's law fits these values with a parallel fraction $f$ between $0.75$ and $0.85$: extrapolating this fitted curve gives an asymptote of $4$ to $7$. Communication, scheduling and load imbalance can change with worker count, so the measurements establish performance only over the tested range. More workers also means more copies of the data in memory.

### A reproducible pipeline with targets (★★)

A pipeline declares the steps of an analysis and their dependencies; `targets` only reruns the steps whose code or inputs changed, and can run independent steps in parallel (with `crew`).

**Question** Write a `_targets.R` file with three targets (read the Parquet data, fit the model, compute its RMSE), run it twice with `tar_make()` in a temporary folder, and check that nothing is recomputed the second time. Then change the model and run it again.

**Answer**

````r
pipeline_dir <- file.path(tempdir(), "flights_pipeline")
dir.create(pipeline_dir, showWarnings = FALSE)
file.copy("data/flights_parquet", pipeline_dir, recursive = TRUE)
write_pipeline <- function(formula) writeLines(c(
  'library(targets)',
  'tar_option_set(packages = c("arrow", "dplyr", "tidyr"))',
  'list(',
  '  tar_target(data, arrow::open_dataset("flights_parquet") |> select(ARRIVAL_DELAY, DEPARTURE_DELAY, SCHEDULED_TIME) |> collect() |> drop_na()),',
  sprintf('  tar_target(model, lm(%s, data = data)),', formula),
  '  tar_target(rmse, sqrt(mean(residuals(model)^2)))',
  ')'), file.path(pipeline_dir, "_targets.R"))
run <- function() callr::r(\(dir) { setwd(dir); targets::tar_make(reporter = "silent"); targets::tar_meta(fields = c("time", "seconds")) },
                           args = list(pipeline_dir))
write_pipeline("ARRIVAL_DELAY ~ DEPARTURE_DELAY")
run_1 <- run(); run_2 <- run()
write_pipeline("ARRIVAL_DELAY ~ DEPARTURE_DELAY + SCHEDULED_TIME")
run_3 <- run()
list(`first run` = run_1 |> select(name, time), `second run` = run_2 |> select(name, time), `after changing the model` = run_3 |> select(name, time))
````

**Conclusion** The first `tar_make()` runs the three targets; the second one recomputes nothing (the times of the targets do not change): `targets` knows that neither their code nor their inputs changed. After changing the formula of the model, only the model and its RMSE are recomputed, not the reading of the data. On a long pipeline, this saves hours, and guarantees that the results correspond to the current code.

### In-database scoring with orbital (★★)
